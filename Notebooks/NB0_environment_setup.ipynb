## MASTER ACTIVATOR TEMPLATE

```python

import sys
import os
from google.colab import drive

# 1. Bind Google Drive Environment
drive.mount('/content/drive', force_remount=False)

# 2. Establish Project Paths
LIB_DIR  = "/content/drive/MyDrive/radiology_ai/libs"
DATA_DIR = "/content/drive/MyDrive/radiology_ai/data"
MDL_DIR  = "/content/drive/MyDrive/radiology_ai/models"
RES_DIR  = "/content/drive/MyDrive/radiology_ai/results"
RAG_DIR  = "/content/drive/MyDrive/radiology_ai/rag_papers"
CHR_DIR  = "/content/drive/MyDrive/radiology_ai/chroma_db"

for d in [LIB_DIR, DATA_DIR, MDL_DIR, RES_DIR, RAG_DIR, CHR_DIR]:
    os.makedirs(d, exist_ok=True)

# 3. 🛡️ NON-DESTRUCTIVE ISOLATION: Temporarily mask the Drive path from Python's search sequence
sys.path = [p for p in sys.path if p != LIB_DIR]

# 4. Flush tracking RAM elements out of memory cache
_stale = ['chromadb', 'gradio', 'sentence_transformers', 'pydantic',
          'huggingface_hub', 'langchain', 'transformers', 'albumentations']
for m in list(sys.modules.keys()):
    if any(s in m for s in _stale):
        del sys.modules[m]

# 5. Native Load Verification (Ensures baseline tools stay clean)
import transformers
print(f"📦 System Core Engine: transformers v{transformers.__version__} bound natively from local memory ✅")

# 6. Append Drive libs as low-priority pathing for downstream scripts
if LIB_DIR not in sys.path:
    sys.path.append(LIB_DIR)

# 7. Compute Target Verification
import torch
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print(f"✅ Environment Configured Safely | Target Device: {DEVICE.upper()}")
if DEVICE == "cuda":
    print(f"🚀 Active HW Accelerator: {torch.cuda.get_device_name(0)} ({torch.cuda.get_device_properties(0).total_memory/1e9:.1f} GB VRAM)")
else:
    print("⚠️ No GPU detected. Adjust settings via Runtime -> Change runtime type -> T4 GPU")
```
